```python
courses = ['Finance', 'Mathematics', 'Python', 'Excel', 'English']
print(courses)

print(courses[1]) #You can index/list individual items by this

print(courses[-1]) #Last item of the list

print(courses[0:3]) # First 3 items of the list

print(courses[2:]) # Starting items from 3rd item

# Inserting items to the list

courses.insert(3,'SQL') #Adding item to a specific position. Eg: adding SQL to the 4th position
print(courses)

courses.append('Quant')  #Without specific location
print(courses)

courses_2 = ['Internships', 'Projects']

# Adding another list to a specific position
courses.insert(7,courses_2)  #This will create a list within the list, also don't use ' ' here otherwise it will only add the name of the list
print(courses)

final = ['Job']
courses.extend(final)  #This will actually add the items to the original list without making another list
print(courses)

# Removing things

courses.remove('English')
print(courses)

#Removing the last item:

courses.append('CompSci')
print(courses)

popped = courses.pop()
print(popped) # Will show the removed item (last one), used in stacks and queues
print(courses)

```
