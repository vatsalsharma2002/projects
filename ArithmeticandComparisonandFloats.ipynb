```python
#Addition = 2 + 3
#Subtraction = 3-2
#Multiplication = 2*3
#Division = 3/2 (will give remainder till decimal)
#Floor Division = 3//2
#Squaring something = 3**2 (will give remainder as whole number only)
#Modulus = 3%2 (will give remainder)

# BODMAS rule is followed here

print(3+2)

print(3-2)

print(3*2)

print(3/2)

print(3//2)

print(3**2)


#BODMAS EXAMPLES

print(3*2+1)

print(3*(2+1))

num = 1

num = num+1
print(num)

#or you can do this too

num2 = 2

num2 +=1

print(num2)

# you can perform similar operations by just putting '=' sign after an operation

num3 = 3
num3 -= 1

print(num3)

num4 = 4
num4 *= 10
print(num4)


num5 = 10
num5 /=5
print(num5)

# Absolute values (mod)

print(abs(-3))

# Rounding off numbers

print(round(3.75))

print(round(3.7575, 1))  #rounding off till 1 digit
print(round(3.9876,2)) #rounding off till 2 digits

#Comparisons

# Equal: 3==2   (we use double equals here because single equal is used to take input)
# Not Equal: 3!=2
# Greater than: 3>2
# Less than: 2<3
# Greater than equal to: 3>=2
# Less than equal to: 2<=3

num_a = 2
num_b = 3

print(num_a == num_b)
print(num_a != num_b)
print(num_a>num_b)
print(num_a<num_b)
print(num_a>=num_b)
print(num_a<=num_b)

# Making floats(strings that can be added together)

x = 100
y = 200

x = int(x)
y = int(y)

print(x+y)
```
