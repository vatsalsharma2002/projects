```python
courses = ['Finance', 'Python', 'Maths', 'SQL']

# Reverses list order
courses.reverse()
print(courses)

# Arranges according to alphabetical order
courses.sort()
print(courses)

# Check index or position of an item

print(courses.index('Finance'))

# Checking if an item is in the list or not:

print('English' in courses)
print('Maths' in courses)

nums = [1,5,4,2,3]
nums.sort()   #Ascending order
print(nums)


# For descending order:

courses.sort(reverse=True)
nums.sort(reverse=True)

print(courses)
print(nums)


# To get the sorted version without altering the original list

sorted_courses = sorted(courses)
print(sorted_courses)


# Min, Max, Sum

print(min(nums))
print(max(nums))
print(sum(nums))

#Looping values:

for item in courses:
  print(item)

# To get items and index

for item in enumerate(courses):
  print(item)

# Second way:

for index, item in enumerate(courses):
  print(index, item)

# To assign a value to items:

for index, item in enumerate(courses, start=2):
  print(index, item)




# Splitting values:

courses_str = ','.join(courses)
print(courses_str)

courses_str1 = '-'.join(courses)
print(courses_str1)

new_list = courses_str.split('_')
print(new_list)


# TUPLES: values cannot be added or removed, uses () not square brackets like lists

tuple_eg = ('Finance', 'Python', 'Maths', 'SQL')
print(tuple_eg)


# SETS: work just like their mathematical definition, uses {}

set_eg = {'Math', 'Math', 'Finance', 'Python'}   # added repetition to give an example
print(set_eg)

# Intersection, Union, Compliment of sets:

cs_courses = {'CompSci','History','Math'}
art_courses = {'History','Arts'}

print(cs_courses.intersection(art_courses))

print(cs_courses.union(art_courses))

print(cs_courses.difference(art_courses))


# Empty List

empty_list = []
empty_list = list()

# Empty Tuple

empty_tuple = ()
empty_tuple = tuple()

# Empty Set

empty_set = {} # THIS IS WRONG -> This is a dict not an empty set
empty_set = set() # this is an empty set
```
